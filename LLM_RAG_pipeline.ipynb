# Mitigating Mistakes using a RAG System

```python
!pip install -q langchain langchain-community langchain-ollama langchain-text-splitters chromadb beautifulsoup4 lxml
```

```python
#text scraping
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

url = "https://en.wikipedia.org/wiki/Los_Gatos,_California#:~:text=Los%20Gatos%20is%20Spanish%20for%20%22The%20Cats%22."
loader = WebBaseLoader(url)
docs = loader.load()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
splits = text_splitter.split_documents(docs)

print(f"Loaded {len(docs)} document(s)")
print(f"Created {len(splits)} chunks")
```

```python
#create embeddings
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="nomic-embed-text")

vectorstore = Chroma.from_documents(
    documents=splits,
    embedding=embeddings
)

retriever = vectorstore.as_retriever(search_kwargs={"k": 4})
```

```python
#RAG function
def rag_answer(question):
    retrieved_docs = retriever.invoke(question)
    context = "\n\n".join(doc.page_content for doc in retrieved_docs)

    prompt = f"""
Use only the context below to answer the question as accurately as possible.
If the answer is not supported by the context, say so.

Context:
{context}

Question:
{question}

Answer:
"""

    response = llm.invoke(prompt)
    return response, retrieved_docs
```

```python
#asking the incorrect questions from part b
question = "When was the Town of Los Gatos founded?"
response, retrieved_docs = rag_answer(question)
print(response)

question = "Which major companies, if any, have currently set up headquarters in Los Gatos?"
response, retrieved_docs = rag_answer(question)
print(response)

question = "What is the population of the Town of Los Gatos?"
response, retrieved_docs = rag_answer(question)
print(response)

question = "Is Los Gatos situated in California's wine country?"
response, retrieved_docs = rag_answer(question)
print(response)
```

I used the town's Wikipedia page instead of the official .ca website because Wikipedia has all its information on one page, so the loader can see most of the town information on the page HTML. The __facts are now more accurate compared to previous answers.__ Three of the four questions are answered correctly/satisfactorily. I appreciate the model's ability to follow instructions and NOT generate information it does not have access to, thus not answering the question related to the date of the Town's founding. This type of response is way better than the model reproducing incorrect text or making oversimplifications. There is still some inaccuracy when it comes to Los Gatos' location in proximity to Wine Country, though at this point, I suspect it's a matter of semantics.
